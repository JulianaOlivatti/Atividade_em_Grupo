```python
# Verificar o tamanho de conta a pagar

valor = float(input("informe o valor a pagar: "))

if valor < 10:
    print("O valor a pagar é menor do que 10")
elif valor > 10:
    print("O valor a pagar é maior do que 10")
else:
    print("O valor a pagar é igual a 10")
```

```python
# Agora queremos que o usuário informe todos os valores
# de uma comanda

# Laço - wile (enquanto) - ter condição ou não
# Se não tem condição, eu tenho que falar quando termina

saldo = 0

while True: # Funciona enquanto o sistema for verdadeiro
  valor = float(input("Informe o valor do item \n"))
  saldo = saldo + valor

  pergunta = input("Tem um novo valor a informar? \n")

  if pergunta.lower() == "sim" or pergunta == "s":
    print("Informe um novo valor \n")
  else:
    print(f" O seu saldo é de {saldo} \n")
    break


```

```python
# Quero cadastrar o produto consumido

list_produtos = []

while True:
  produto = input("Informe o nome do produto: \n")

# Como colocar o valor dentro da lista
  list_produtos.append(produto)

  pergunta = input("Tem um novo produto a informar? \n")

  if pergunta.lower() == "sim" or pergunta == "s":
    print("Informe um novo produto \n")
  else:
    # print(list_produtos) gera uma vizualização desagradavel neste caso
    # O comando FOR percorre uma estrutura inteira
    # Como funciona?
    # FOR VARIAVEL in ESTRUTURA

    for i in list_produtos:
      print(i)

    break
```

```python
list_produtos = []
list_valores = []

while True:
    produto = input('Informe o nome do produto:\n')
    valor = float(input(f'Informe o valor para {produto}\n'))

    list_produtos.append(produto)
    list_valores.append(valor)

    pergunta = input('Deseja cadastrar um novo produto?\n')

    if pergunta.lower() == 'sim' or pergunta.lower() == 's':
        print("Vamos iniciar um novo cadastro")
    else:
      # i - novo produto
      # j - valor
        for i,j in zip(list_produtos, list_valores):
          # Zip -- Juntar um unico arquivo
          # produto -- R$ valor
          print(f"O produto {i} custa R$ {j}")

        print(f"\n\n\n Total da compra {sum(list_valores)}")
        # O comando SUM(lista) soma todos os valores de dentro
        # de uma lista
        # O comando SUM(lista, valor) faz com que ele some
        # valor que informamos também

        taxa = sum(list_valores)*0.1
        print(f"Valor com taxa de serviço {sum(list_valores,taxa)}")

        break
```

```python
while True:
    entrada = input("Digite um número ou 'sair': ")

    if entrada.lower() == "sair":
        print("Programa encerrado.")
        break

    if not entrada.isdigit():
        print("Digite um número inteiro válido.")
        continue

    n = int(entrada)

    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)


```
