```python
notaA = float(input("Informe a primeira nota: "))
notaB = float(input("Informe a segunda nota: "))

mediafinal = (notaA + notaB) / 2

if mediafinal >= 7.0:
    print(f"A média: {mediafinal:.1f} - Aprovado")
else:
    print(f"A média: {mediafinal:.1f} - Reprovado")
```
